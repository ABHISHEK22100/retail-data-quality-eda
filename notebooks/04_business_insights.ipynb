# Business Insights & Recommendations

## 1. Project Objective

The objective of this analysis is to evaluate retail store sales performance, 
identify important business patterns, and provide actionable recommendations 
based on the available transaction data.

### Key Business Areas
- Product Category Performance
- Payment Method Analysis
- Sales Channel Performance
- Monthly and Yearly Revenue Trends
- Product Demand Analysis
- Customer Revenue Contribution

## 2. Key Business Findings

### 2.1 Product Category Performance
- Butchers generated the highest revenue: ₹208,118.
- Milk Products generated the lowest revenue: ₹180,112.
- Furniture recorded the highest quantity sold: 8,462 units.

### 2.2 Payment Method Analysis
- Cash was the most frequently used payment method.
- Cash generated the highest revenue: ₹537,710.

### 2.3 Sales Channel Performance
- Online transactions generated ₹791,401.
- In-store transactions generated ₹760,670.
- Online sales contributed more revenue than in-store sales.

### 2.4 Sales Trend
- 2024 generated the highest revenue among complete calendar years.
- Revenue declined in 2023 compared with 2022, followed by recovery in 2024.
- January 2025 contains partial-month data.

### 2.5 Customer Analysis
- All 25 identified customers made repeat purchases.
- Top 10 customers contributed 41.82% of total revenue.

## 3. Business Recommendations

### 3.1 Product Category Strategy
- Monitor Butchers category performance to understand its higher revenue contribution.
- Investigate why Milk Products generate comparatively lower revenue.
- Maintain appropriate inventory levels for high-demand categories.

### 3.2 Payment Strategy
- Continue supporting Cash payments.
- Improve the digital payment experience to encourage adoption of digital payment methods.

### 3.3 Sales Channel Strategy
- Maintain the online sales channel because it currently generates higher revenue.
- Investigate opportunities to improve in-store sales performance.

### 3.4 Customer Retention
- Consider loyalty programs for high-value customers.
- Monitor customer purchasing patterns to identify changes in buying behaviour.

### 3.5 Sales Monitoring
- Track monthly revenue to identify unusual increases or decreases.
- Compare complete calendar years rather than comparing partial-year data with full-year data.

## 4. Data Limitations

1. The dataset contains only 25 unique customers, which limits customer-level generalization.
2. Missing Quantity and Total Spent values remain unresolved for 604 transactions.
3. Discount Applied contains 4,199 missing values.
4. Missing product names were replaced with "Unavailable", limiting product-level demand analysis.
5. January 2025 contains partial-month data and is not directly comparable with complete months.
6. The dataset does not provide sufficient information to establish the causes of revenue changes.
