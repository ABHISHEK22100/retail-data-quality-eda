```python
numeric_columns = [
    "Price Per Unit",
    "Quantity",
    "Total Spent"
]

display(df[numeric_columns].describe().T)
```

```python
for col in numeric_columns:
    print(f"\n--- {col} ---")
    print("Missing:", df[col].isna().sum())
    print("Zero values:", (df[col] == 0).sum())
    print("Negative values:", (df[col] < 0).sum())
```

```python
# Inspect rows where all three numeric fields are available

audit_df = df[
    ["Transaction ID", "Price Per Unit", "Quantity", "Total Spent"]
].dropna().copy()

audit_df["Expected_Total"] = (
    audit_df["Price Per Unit"] * audit_df["Quantity"]
)

audit_df["Difference"] = (
    audit_df["Total Spent"] - audit_df["Expected_Total"]
)

display(audit_df.head(10))
```

```python
print("Rows checked:", len(audit_df))

print(
    "Exact matches:",
    np.isclose(
        audit_df["Total Spent"],
        audit_df["Expected_Total"],
        atol=0.01
    ).sum()
)

print(
    "Non-matching rows:",
    (~np.isclose(
        audit_df["Total Spent"],
        audit_df["Expected_Total"],
        atol=0.01
    )).sum()
)
```
